# ATDL Final-One — Tasks 5–12, verified production execution

## 0. Project paths and immutable controls

```python
from pathlib import Path
import json, subprocess, sys, torch
ROOT = Path('/home/vu-lab03-pc17/ATDL-1').resolve()
assert ROOT.exists()
PATHS = {
 'teacher_t1': ROOT/'resnet34_cifar10_fp32_best.pth',
 'student_t2': ROOT/'results/best_models/task2_b1_fp32_resnet18_best.pth',
 'ternary_t3': ROOT/'results/best_models/task3_b2_default_best.pth',
 't4_aggregate': ROOT/'results/task4/task4_b3_final_t2_lam09_aggregate_summary.json',
 't4_best': ROOT/'results/task4/best_models/task4_b3_final_t2_lam09_remaining_rerun1_best.pth',
 'dkd': ROOT/'results/task4/task5_dkd_final_t2_lam09_a1_b8_r1_summary.json',
 'dist': ROOT/'results/task4/task7_dist_t1_lam05_r1_summary.json',
 'trainer': ROOT/'scripts/train_student_vanilla_kd.py',
 'pipeline': ROOT/'scripts/run_final_notebook_pipeline.py',
 'data_module': ROOT/'src/data.py',
}
assert all(p.exists() for p in PATHS.values()), [str(p) for p in PATHS.values() if not p.exists()]
for name,path in PATHS.items(): print(f'{name:14} {path.relative_to(ROOT)}')
```

## 1. Frozen split, teacher/student identity, and test firewall

```python
teacher_ckpt = torch.load(PATHS['teacher_t1'], map_location='cpu', weights_only=False)
student_ckpt = torch.load(PATHS['student_t2'], map_location='cpu', weights_only=False)
t4 = json.loads(PATHS['t4_aggregate'].read_text())
assert teacher_ckpt['arch'] == 'ResNet34-CIFAR' and teacher_ckpt['num_classes'] == 10
assert student_ckpt['arch'] == 'ResNet18-CIFAR' and student_ckpt['num_classes'] == 10
assert t4['training_seeds'] == [42, 43, 44] and t4['test_evaluation'] == 'not_run'
for source in [PATHS['trainer'], PATHS['pipeline']]:
    hits = [x for x in ('get_test_loader(', 'CIFAR10Test', 'test_loader =') if x in source.read_text()]
    assert not hits, f'Test-firewall violation in {source}: {hits}'
print({'dataset':'CIFAR-10', 'split':'Task1 classwise 45k/5k seed 42', 'teacher':'T1 ResNet34', 'student':'T2 ResNet18 warm-start', 'task4_mean':t4['validation_mean'], 'test':'LOCKED'})
```

## 2. Completed Tasks 5–6: advanced logits and quantized targets

DKD and DIST are preserved completed artifacts. QFD-style quantized-feature distillation is already a completed one-seed screen. QTRD is the remaining production method from this family: it transfers pairwise geometry of a frozen ternarized teacher terminal feature.

```python
EXECUTABLE_SCREENS = [
 {'task':'T6', 'method':'qfd', 'config':'configs/kd/resnet18_ternary_qfd_screen.yaml', 'run':'task8_qfd_screen_t2_lam09_aux1_r1'},
 {'task':'T7', 'method':'at', 'config':'configs/kd/resnet18_ternary_at_screen.yaml', 'run':'task9_at_screen_t2_lam09_aux1_r1'},
 {'task':'T8', 'method':'rkd', 'config':'configs/kd/resnet18_ternary_rkd_screen.yaml', 'run':'task10_rkd_screen_t2_lam09_aux1_r1'},
 {'task':'T6', 'method':'qtrd', 'config':'configs/kd/resnet18_ternary_qtrd_screen.yaml', 'run':'task11_qtrd_screen_t2_lam09_aux1_r1'},
]
for stage in EXECUTABLE_SCREENS:
    assert (ROOT/stage['config']).exists(), stage
    print(stage)
# Each stage preserves teacher/split/optimizer/QAT and differs only in the loss signal.
```

## 3. Tasks 7–8: attention and relational transfer

AT transfers normalized terminal attention maps; RKD transfers scale-normalized pairwise penultimate geometry. Their existing one-seed results are retained. No local duplicate model or quantizer is used.

```python
def result_path(run): return ROOT/'results/task4'/f'{run}_summary.json'
for stage in EXECUTABLE_SCREENS[:-1]:
    p = result_path(stage['run'])
    assert p.exists(), f'Completed stage artifact missing: {p}'
    d = json.loads(p.read_text())
    assert d['test_evaluation'] == 'not_run' and d['all_conv_and_fc_ternary']
    print(stage['method'], d['validation_mean'], d['validation_best'], 'validation-only')
```

## 4. Training — resumable QTRD one-seed screen

This runs the shared artifact-first controller. Existing QFD, AT and RKD summaries are skipped. QTRD receives a 1-epoch smoke test and then a 50-epoch seed-42 screen. The controller fails closed on any test-firewall, reload, or ternary-verification failure.

```python
RUN_PIPELINE = True
cmd = [sys.executable, str(PATHS['pipeline'])]
print('Executing:', ' '.join(cmd))
if RUN_PIPELINE:
    done = subprocess.run(cmd, cwd=ROOT)
    assert done.returncode == 0, 'Pipeline failure: preserve artifacts and inspect logs before retrying.'
```

## 5. Task 9: diagnostic integrity

Saved histories provide CE/KD contributions, gradient norms/cosines, teacher/student agreement, entropy, confidence, sparsity, and layerwise quantization metrics. Representation similarity was not logged in pre-existing runs and is not fabricated.

```python
diagnostic_runs = ['task4_b3_final_t2_lam09_rerun1', 'task5_dkd_final_t2_lam09_a1_b8_r1', 'task7_dist_t1_lam05_r1']
for run in diagnostic_runs:
    files = sorted((ROOT/'experiments/task4/diagnostics'/run).glob('seed*.json'))
    print(run, [f.relative_to(ROOT).as_posix() for f in files])
```

## 6. Task 10: ablation evidence

The existing T×λ Task-4 screening stages are the controlled vanilla-KD ablation. Do not rerun a new arbitrary grid after seeing the final validation results.

```python
ablation_summaries = sorted((ROOT/'results/task4').glob('task4_tlambda_*_summary.json'))
print('Existing controlled T×λ ablations:', len(ablation_summaries))
for p in ablation_summaries[:6]:
    x=json.loads(p.read_text()); print(p.name, x['temperature'], x['kd_lambda'], x['validation_mean'])
```

## 7. Tasks 11–12: compression and final selection

The required final selection rule is validation mean ± standard deviation, seed stability, strict ternary correctness, reload integrity, and quantization diagnostics. A one-seed screen is provisional and cannot displace the Task-4 three-seed control.

```python
all_results = {'T4_vanilla_3seed': PATHS['t4_aggregate'], 'DKD_3seed': PATHS['dkd'], 'DIST_1seed': PATHS['dist']}
for stage in EXECUTABLE_SCREENS:
    p = result_path(stage['run'])
    if p.exists(): all_results[stage['method'].upper()+'_1seed'] = p
table = {}
for name,p in all_results.items():
    x=json.loads(p.read_text()); table[name] = {'mean':x['validation_mean'], 'std':x['validation_std'], 'best':x['validation_best'], 'test':x['test_evaluation']}
print(json.dumps(table, indent=2))
assert all(row['test']=='not_run' for row in table.values())
print('Current rigorous final control: Task 4 vanilla KD, 95.12% ± 0.15% validation.')
```
