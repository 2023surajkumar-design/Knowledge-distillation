# ATDL Assignment — Tasks 5→12 : Advanced Knowledge Distillation for a Ternary ResNet18

**Knowledge transfer under a discrete capacity constraint.** This single notebook carries the project
from the already-trained Task 1–4 checkpoints through every remaining task, with visualizations and
interpretation at each stage:

| Part | Task(s) | Content |
|---|---|---|
| 0–2 | infra | config, shared code (data/models/ternary/STE), load teacher + establish the B0–B3 ladder |
| 3 | **T5** | Advanced logit KD — **DKD**, **DIST** |
| 4 | **T5** | Feature & relation KD — **FitNet**, **Attention Transfer**, **RKD** |
| 5 | **T6** | Quantization-aware transfer — **QFD** (quantized target), **QTRD** (quantized relational) |
| 6 | **T7** | STE study — identity vs clipped vs scaled; gradient diagnostics |
| 7 | **T8** | Progressive ternarization (FP-warmup) — falsifiable test |
| 8 | **T9** | Adaptive KD — teacher-confidence-weighted transfer |
| 9 | **T10** | Gradient-conflict analysis — cos(g_CE, g_KD) |
| 10 | **T11** | Consolidated ablation matrix |
| 11 | — | Representation analysis (CKA, penultimate geometry) |
| 12 | — | Layer sensitivity (per-layer quant error / sparsity / α) |
| 13 | **T12** | Compression analysis (honest theoretical vs real) |
| 14 | — | **Final test evaluation** (firewalled, once) + calibration/confusion |
| 15 | — | Summary, best method, scientific conclusion, limitations |

**Central hypothesis** (carried from the research plan): the limitation is *representational*, not
informational — a ternary student cannot reproduce the teacher's continuous FP32 feature geometry, so
forcing exact imitation supplies gradients it can never zero out. The productive paradigm is to
**transform teacher knowledge into an attainable, relation-preserving target** inside the ternary
function class. Parts 3–5 test this directly; the pivotal empirical check is **no-KD vs vanilla-KD**.

> **Scientific integrity.** This notebook *computes* every number when you run it — no accuracy is
> hard-coded. Interpretation cells tell you how to read the numbers you get, including how to report an
> honest negative result (e.g. if KD hurts the ternary student). Only the final cell touches the test set.

## Part 1 — Shared Infrastructure

Self-contained, matching the Task 1–4 code exactly so your saved teacher/student checkpoints load
without surgery. Frozen 45k/5k stratified split (seed 42) + the **test-set firewall** (test is only
reachable with `final_evaluation=True`, used once in Part 14).

```python
# ---------------- data (frozen split + firewall) ----------------
CIFAR10_CLASSES = ["airplane","automobile","bird","cat","deer","dog","frog","horse","ship","truck"]
MEAN, STD = (0.4914,0.4822,0.4465), (0.2470,0.2435,0.2616)
def _tf():
    tr = transforms.Compose([transforms.RandomCrop(32,padding=4), transforms.RandomHorizontalFlip(),
                             transforms.ToTensor(), transforms.Normalize(MEAN,STD)])
    ev = transforms.Compose([transforms.ToTensor(), transforms.Normalize(MEAN,STD)])
    return tr, ev

def stratified_split(targets, val_fraction=0.1, num_classes=10, seed=42):
    targets = np.asarray(targets); vpc = int(len(targets)*val_fraction/num_classes)
    g = torch.Generator().manual_seed(seed); tr, va = [], []
    for c in range(num_classes):
        idx = np.flatnonzero(targets==c)
        perm = idx[torch.randperm(len(idx), generator=g).numpy()]
        va += perm[:vpc].tolist(); tr += perm[vpc:].tolist()
    assert len(tr)+len(va)==len(targets) and set(tr).isdisjoint(va)
    return tr, va

def get_loaders(batch=128, workers=4, root="./data", seed=42):
    tr_tf, ev_tf = _tf()
    base_tr = torchvision.datasets.CIFAR10(root, train=True, download=True, transform=tr_tf)
    base_va = torchvision.datasets.CIFAR10(root, train=True, download=True, transform=ev_tf)
    tri, vai = stratified_split(base_tr.targets)
    pin = torch.cuda.is_available(); pw = workers>0
    tl = DataLoader(Subset(base_tr,tri), batch_size=batch, shuffle=True, num_workers=workers, pin_memory=pin, persistent_workers=pw)
    vl = DataLoader(Subset(base_va,vai), batch_size=batch, shuffle=False, num_workers=workers, pin_memory=pin, persistent_workers=pw)
    return tl, vl

def get_test_loader(batch=128, workers=4, root="./data", final_evaluation=False):
    if not final_evaluation:
        raise RuntimeError("TEST SET LOCKED (C9/C11): pass final_evaluation=True only for the final report.")
    _, ev_tf = _tf()
    ds = torchvision.datasets.CIFAR10(root, train=False, download=True, transform=ev_tf)
    return DataLoader(ds, batch_size=batch, shuffle=False, num_workers=workers, pin_memory=torch.cuda.is_available())
print("data helpers ready")
```

```python
# ---------------- models (CIFAR ResNet18/34, checkpoint-compatible, feature-returning) ----------------
class BasicBlock(nn.Module):
    expansion = 1
    def __init__(self, cin, cout, stride=1):
        super().__init__()
        self.conv1 = nn.Conv2d(cin, cout, 3, stride, 1, bias=False); self.bn1 = nn.BatchNorm2d(cout)
        self.conv2 = nn.Conv2d(cout, cout, 3, 1, 1, bias=False);      self.bn2 = nn.BatchNorm2d(cout)
        self.shortcut = nn.Sequential()
        if stride != 1 or cin != cout:
            self.shortcut = nn.Sequential(nn.Conv2d(cin, cout, 1, stride, bias=False), nn.BatchNorm2d(cout))
    def forward(self, x):
        o = F.relu(self.bn1(self.conv1(x))); o = self.bn2(self.conv2(o))
        return F.relu(o + self.shortcut(x))

class ResNetCIFAR(nn.Module):
    def __init__(self, blocks, num_classes=10):
        super().__init__(); self.in_c = 64
        self.conv1 = nn.Conv2d(3,64,3,1,1,bias=False); self.bn1 = nn.BatchNorm2d(64)
        self.layer1 = self._make(64, blocks[0], 1); self.layer2 = self._make(128, blocks[1], 2)
        self.layer3 = self._make(256, blocks[2], 2); self.layer4 = self._make(512, blocks[3], 2)
        self.avgpool = nn.AdaptiveAvgPool2d((1,1)); self.fc = nn.Linear(512, num_classes)
        for m in self.modules():
            if isinstance(m, nn.Conv2d): nn.init.kaiming_normal_(m.weight, mode="fan_out", nonlinearity="relu")
            elif isinstance(m, nn.BatchNorm2d): nn.init.constant_(m.weight,1); nn.init.constant_(m.bias,0)
            elif isinstance(m, nn.Linear): nn.init.normal_(m.weight,0,0.01); nn.init.constant_(m.bias,0)
    def _make(self, cout, n, stride):
        layers=[];
        for s in [stride]+[1]*(n-1): layers.append(BasicBlock(self.in_c, cout, s)); self.in_c = cout
        return nn.Sequential(*layers)
    def forward(self, x, return_features=False):
        o = F.relu(self.bn1(self.conv1(x)))
        f1=self.layer1(o); f2=self.layer2(f1); f3=self.layer3(f2); f4=self.layer4(f3)
        p = torch.flatten(self.avgpool(f4),1); logits=self.fc(p)
        if return_features: return logits, {"f1":f1,"f2":f2,"f3":f3,"f4":f4,"penultimate":p}
        return logits

def resnet18(nc=10): return ResNetCIFAR([2,2,2,2], nc)
def resnet34(nc=10): return ResNetCIFAR([3,4,6,3], nc)
def n_params(m): return sum(p.numel() for p in m.parameters())
print("ResNet18 params:", f"{n_params(resnet18()):,}", "| ResNet34 params:", f"{n_params(resnet34()):,}")
```

```python
# ---------------- ternary quantizer + STE (Task 3 core) ----------------
class TernarizeSTE(torch.autograd.Function):
    @staticmethod
    def forward(ctx, w, delta, clip, mode):
        q = torch.zeros_like(w)
        q = torch.where(w > delta, torch.ones_like(w), q)
        q = torch.where(w < -delta, -torch.ones_like(w), q)
        ctx.save_for_backward(w); ctx.clip = clip; ctx.mode = mode
        return q
    @staticmethod
    def backward(ctx, g):
        (w,) = ctx.saved_tensors
        if ctx.mode == "identity":
            return g, None, None, None
        mask = (w.abs() <= ctx.clip).to(g.dtype)          # clipped (Yin et al.)
        if ctx.mode == "scaled":
            return g * mask * 1.0, None, None, None        # placeholder for grad scaling (set below)
        return g * mask, None, None, None

@dataclass
class QuantConfig:
    threshold: str="twn"; scale: str="derived"; symmetric: bool=True
    scope: str="per_channel"; t: float=0.05; ste: str="clipped"; clip: float=1.0

class TernaryQuantizer(nn.Module):
    def __init__(self, shape, cfg):
        super().__init__(); self.cfg=cfg; self.C=shape[0]; self.nd=len(shape)
        g = self.C if cfg.scope=="per_channel" else 1
        self._view = (g,)+(1,)*(self.nd-1); self.g=g
        self.ternary_enabled = True   # toggled False during FP-warmup (progressive, Part 7)
        if cfg.scale=="learned":
            if cfg.symmetric: self.alpha = nn.Parameter(torch.ones(g))
            else: self.alpha_p = nn.Parameter(torch.ones(g)); self.alpha_n = nn.Parameter(torch.ones(g))
    def _flat(self, w): return w.reshape(self.C,-1) if self.cfg.scope=="per_channel" else w.reshape(1,-1)
    def _delta(self, w):
        f = self._flat(w).abs()
        d = self.cfg.t*f.max(1).values if self.cfg.threshold=="ttq" else 0.7*f.mean(1)
        return d.view(self._view)
    def _derived_alpha(self, w, delta):
        f=self._flat(w).abs(); d=delta.reshape(self.g,1); m=(f>d).float()
        return ((f*m).sum(1)/m.sum(1).clamp(min=1.0)).view(self._view)
    @torch.no_grad()
    def init_learned(self, w):
        if self.cfg.scale!="learned": return
        a = self._derived_alpha(w, self._delta(w)).reshape(self.g).clamp(min=1e-3)
        if self.cfg.symmetric: self.alpha.copy_(a)
        else: self.alpha_p.copy_(a); self.alpha_n.copy_(a)
    def forward(self, w):
        if not self.ternary_enabled: return w   # FP pass-through (progressive warmup)
        delta = self._delta(w)
        q = TernarizeSTE.apply(w, delta, self.cfg.clip, self.cfg.ste)
        if self.cfg.scale=="derived": return q * self._derived_alpha(w, delta).detach()
        if self.cfg.symmetric: return q * self.alpha.view(self._view)
        sel = torch.where(q>0, self.alpha_p.view(self._view), self.alpha_n.view(self._view))
        return q * sel
    @torch.no_grad()
    def stats(self, w):
        delta=self._delta(w); q=torch.zeros_like(w)
        q=torch.where(w>delta,torch.ones_like(w),q); q=torch.where(w<-delta,-torch.ones_like(w),q)
        wh=self.forward(w); n=w.numel()
        return dict(n=n, neg=(q<0).sum().item()/n, zero=(q==0).sum().item()/n, pos=(q>0).sum().item()/n,
                    delta=delta.mean().item(), qerr=(w-wh).norm().item()/(w.norm().item()+1e-12),
                    alpha=self._derived_alpha(w,delta).mean().item() if self.cfg.scale=="derived"
                          else (self.alpha.mean().item() if self.cfg.symmetric else self.alpha_p.mean().item()))

class TernaryConv2d(nn.Conv2d):
    def __init__(self,*a,quant_cfg=None,**k): super().__init__(*a,**k); self.quant=TernaryQuantizer(self.weight.shape, quant_cfg or QuantConfig())
    def forward(self,x): return self._conv_forward(x, self.quant(self.weight), self.bias)
    @torch.no_grad()
    def deployed_weight(self): return self.quant(self.weight)
class TernaryLinear(nn.Linear):
    def __init__(self,*a,quant_cfg=None,**k): super().__init__(*a,**k); self.quant=TernaryQuantizer(self.weight.shape, quant_cfg or QuantConfig())
    def forward(self,x): return F.linear(x, self.quant(self.weight), self.bias)
    @torch.no_grad()
    def deployed_weight(self): return self.quant(self.weight)

def convert_to_ternary(model, cfg, keep_first_last_fp32=False):
    reps=[]
    for name,m in model.named_modules():
        if isinstance(m,(nn.Conv2d,nn.Linear)) and not isinstance(m,(TernaryConv2d,TernaryLinear)):
            if keep_first_last_fp32 and (name.endswith("conv1") and "layer" not in name or name.endswith("fc")): continue
            reps.append((name,m))
    for name,m in reps:
        parent=model; *ps,attr=name.split(".")
        for p in ps: parent=getattr(parent,p)
        if isinstance(m,nn.Conv2d):
            nw=TernaryConv2d(m.in_channels,m.out_channels,m.kernel_size,stride=m.stride,padding=m.padding,
                             dilation=m.dilation,groups=m.groups,bias=(m.bias is not None),quant_cfg=cfg)
        else:
            nw=TernaryLinear(m.in_features,m.out_features,bias=(m.bias is not None),quant_cfg=cfg)
        nw.weight.data.copy_(m.weight.data)
        if m.bias is not None: nw.bias.data.copy_(m.bias.data)
        if cfg.scale=="learned": nw.quant.init_learned(nw.weight.data)
        setattr(parent,attr,nw)
    return model

def ternary_param_groups(model, wd):
    dec,nod=[],[]
    for n,p in model.named_parameters():
        if not p.requires_grad: continue
        (dec if (p.ndim>=2 and n.endswith(".weight") and "bn" not in n.lower()) else nod).append(p)
    return [{"params":dec,"weight_decay":wd},{"params":nod,"weight_decay":0.0}]

def set_ternary_enabled(model, flag):
    for m in model.modules():
        if isinstance(m,(TernaryConv2d,TernaryLinear)): m.quant.ternary_enabled=flag

@torch.no_grad()
def verify_ternary(model):
    total=zero=0; ok=True; rows=[]
    for name,m in model.named_modules():
        if isinstance(m,(TernaryConv2d,TernaryLinear)):
            dep=m.deployed_weight(); C=dep.shape[0]; f=dep.reshape(C,-1)
            mu=max(int(torch.unique(f[c]).numel()) for c in range(C))
            lat=int(torch.unique(m.weight.detach()).numel())
            s=m.quant.stats(m.weight.detach()); total+=s["n"]; zero+=int(round(s["zero"]*s["n"]))
            layer_ok=(mu<=3 and lat>3); ok=ok and layer_ok
            rows.append(dict(layer=name,kind=type(m).__name__,max_unique=mu,latent_unique=lat,ok=layer_ok,**s))
    return dict(all_ternary=ok, sparsity=zero/max(1,total), n_layers=len(rows), layers=rows,
                bits=math.log2(3), compression=32/math.log2(3))
print("ternary quantizer ready")
```

## Part 2b — KD Loss Library & the Unified Training Harness

Every KD method is one function `kd_loss(method, ...)`; every experiment goes through one
`train_student(method, ...)` harness so **the only thing that differs between rows is the method**.
This is what makes the comparison table a controlled experiment.

**Feature alignment is free here.** ResNet18 and ResNet34 both end their four stages at
512×4×4 feature maps and a 512-d penultimate vector, so feature/relation/quantized-target losses match
teacher↔student dimensions with **no projection** — we compare the same-shaped `f4` maps and penultimate
vectors directly.

Method hypotheses (each answers *what problem does it solve?*):
- **vanilla** — transfer class-probability structure (Hinton).
- **dkd** — decouple target vs non-target KD so the non-target "dark knowledge" isn't suppressed.
- **dist** — match *correlations* not exact logits: robust when the teacher is much stronger.
- **feature (FitNet)** — transfer intermediate representation (raw `f4` MSE).
- **at (Attention Transfer)** — transfer *where* the teacher looks (spatial attention of `f4`).
- **rkd** — transfer *geometry between samples* (pairwise distance + angle), scale/space invariant.
- **qfd** — match the teacher's `f4` **after quantizing it through the student's own quantizer** (an *attainable* target).
- **qtrd** — match the *correlation structure* of a **quantized** teacher penultimate (attainable + relational). *The flagship of the central hypothesis.*

```python
# ---------------- KD losses ----------------
def _kl_T(s_logits, t_logits, T):
    return F.kl_div(F.log_softmax(s_logits/T,1), F.softmax(t_logits/T,1), reduction="batchmean")*(T*T)

def dkd_loss(s, t, y, T, alpha=1.0, beta=8.0):
    # Decoupled KD (Zhao et al. 2022): TCKD (target vs non-target binary) + NCKD (non-target dist).
    gt = F.one_hot(y, s.size(1)).bool()
    ps = F.softmax(s/T,1); pt = F.softmax(t/T,1)
    # binary target/non-target
    def cat(p):
        pt_t = p[gt].reshape(-1,1); pt_n = p.masked_fill(gt,0).sum(1,keepdim=True); return torch.cat([pt_t,pt_n],1)
    bs, bt = cat(ps), cat(pt)
    tckd = F.kl_div(torch.log(bs+1e-7), bt, reduction="batchmean")*(T*T)
    # non-target KD (mask out the target class)
    ls_n = F.log_softmax(s/T - 1000.0*gt, 1); pt_n = F.softmax(t/T - 1000.0*gt, 1)
    nckd = F.kl_div(ls_n, pt_n, reduction="batchmean")*(T*T)
    return alpha*tckd + beta*nckd

def _pearson(a, b, dim):
    a = a - a.mean(dim, keepdim=True); b = b - b.mean(dim, keepdim=True)
    return (a*b).sum(dim) / (a.norm(dim=dim)*b.norm(dim=dim)+1e-8)
def dist_loss(s, t, T=1.0):
    # DIST (Huang et al. 2022): inter-class (rows) + intra-class (cols) correlation of softmax.
    ps, pt = F.softmax(s/T,1), F.softmax(t/T,1)
    inter = (1 - _pearson(ps, pt, dim=1)).mean()
    intra = (1 - _pearson(ps, pt, dim=0)).mean()
    return inter + intra

def feature_loss(fs, ft):                      # FitNet: normalized MSE on f4
    fs = F.normalize(fs.flatten(1), dim=1); ft = F.normalize(ft.flatten(1), dim=1)
    return F.mse_loss(fs, ft)
def attention_loss(fs, ft):                    # AT: match spatial attention maps of f4
    def amap(f): a = f.pow(2).sum(1); return F.normalize(a.flatten(1), dim=1)
    return F.mse_loss(amap(fs), amap(ft))
def rkd_loss(ps, pt, wd=25.0, wa=50.0):        # RKD distance + angle on penultimate
    def pdist(x):
        d = torch.cdist(x, x); return d
    with torch.no_grad():
        dt = pdist(pt); mt = dt[dt>0].mean(); dt = dt/(mt+1e-8)
    ds = pdist(ps); ms = ds[ds>0].mean(); ds = ds/(ms+1e-8)
    ld = F.smooth_l1_loss(ds, dt)
    def angle(x):
        e = (x.unsqueeze(0)-x.unsqueeze(1)); e = F.normalize(e, dim=2)
        return torch.bmm(e, e.transpose(1,2)).view(x.size(0), -1)
    la = F.smooth_l1_loss(angle(ps), angle(pt))
    return wd*ld + wa*la

def _ternarize_feature(ft):                    # quantize teacher feature per-channel (attainable target)
    C = ft.size(1); f = ft.detach()
    flat = f.transpose(0,1).reshape(C,-1)
    delta = 0.7*flat.abs().mean(1).view(1,C,1,1) if f.dim()==4 else 0.7*flat.abs().mean(1).view(1,C)
    q = torch.zeros_like(f); q = torch.where(f>delta,1.0,q); q = torch.where(f<-delta,-1.0,q)
    a = flat.abs().mean(1)
    a = a.view(1,C,1,1) if f.dim()==4 else a.view(1,C)
    return q*a
def qfd_loss(fs, ft):                          # QFD: match student f4 to QUANTIZED teacher f4
    tq = _ternarize_feature(ft)
    return feature_loss(fs, tq)
def qtrd_loss(ps, pt):                         # QTRD: correlation match of quantized teacher penultimate
    ptq = _ternarize_feature(pt)
    inter = (1 - _pearson(ps, ptq, dim=1)).mean()
    intra = (1 - _pearson(ps, ptq, dim=0)).mean()
    return inter + intra

def kd_loss(method, s_out, t_out, y, feats_s, feats_t, T, lam, ce):
    ce_l = ce(s_out, y)
    if method == "none":   return ce_l, ce_l.detach(), torch.zeros((),device=s_out.device)
    if method == "vanilla":kd = _kl_T(s_out, t_out, T)
    elif method=="dkd":    kd = dkd_loss(s_out, t_out, y, T)
    elif method=="dist":   kd = dist_loss(s_out, t_out, T)
    elif method=="feature":kd = feature_loss(feats_s["f4"], feats_t["f4"])
    elif method=="at":     kd = attention_loss(feats_s["f4"], feats_t["f4"])
    elif method=="rkd":    kd = rkd_loss(feats_s["penultimate"], feats_t["penultimate"])
    elif method=="qfd":    kd = qfd_loss(feats_s["f4"], feats_t["f4"])
    elif method=="qtrd":   kd = qtrd_loss(feats_s["penultimate"], feats_t["penultimate"])
    else: raise ValueError(method)
    feature_based = method in ("feature","at","rkd","qfd","qtrd")
    total = ce_l + lam*kd if feature_based else (1-lam)*ce_l + lam*kd
    return total, ce_l.detach(), kd.detach()
print("KD loss library ready")
```

```python
# ---------------- unified training harness ----------------
def warmup_cosine(opt, warm, total):
    def f(e):
        if warm>0 and e<warm: return (e+1)/warm
        p = min(max((e-warm)/max(1,total-warm),0),1); return 0.5*(1+math.cos(math.pi*p))
    return torch.optim.lr_scheduler.LambdaLR(opt, f)

@torch.no_grad()
def evaluate(model, loader, ce):
    model.eval(); L=c=n=0
    for x,y in loader:
        x,y=x.to(device),y.to(device); o=model(x); L+=ce(o,y).item()*y.size(0)
        c+=o.argmax(1).eq(y).sum().item(); n+=y.size(0)
    return L/n, c/n

def build_student(cfg, warm_state=None):
    m = resnet18(10)
    if warm_state is not None: m.load_state_dict(warm_state)
    convert_to_ternary(m, cfg); return m.to(device)

def train_student(method, epochs, seed, loaders, teacher, cfg, *, T=None, lam=None,
                  adaptive=False, fp_warmup=0, ste_override=None, record=True, tag=None, verbose=True):
    set_seed(seed)
    T = CFG["KD_T"] if T is None else T; lam = CFG["KD_LAMBDA"] if lam is None else lam
    qc = copy.deepcopy(cfg)
    if ste_override: qc.ste = ste_override
    tl, vl = loaders
    model = build_student(qc, WARM_STATE)
    ce = nn.CrossEntropyLoss(label_smoothing=CFG["LABEL_SMOOTHING"])
    opt = torch.optim.SGD(ternary_param_groups(model, CFG["WEIGHT_DECAY"]),
                          lr=CFG["LR"]*CFG["BATCH_SIZE"]/128, momentum=CFG["MOMENTUM"], nesterov=CFG["NESTEROV"])
    sch = warmup_cosine(opt, CFG["WARMUP_EPOCHS"], epochs)
    hist = {k:[] for k in ["train_loss","train_acc","val_loss","val_acc","ce","kd","lr","gap","gconf"]}
    best_val, best_state = 0.0, None
    name = tag or f"{method}_T{T:g}_l{lam:g}_s{seed}"
    for ep in range(1, epochs+1):
        set_ternary_enabled(model, ep > fp_warmup)     # progressive FP-warmup (Part 7)
        model.train(); run=ce_r=kd_r=corr=tot=0.0
        it = tqdm(tl, desc=f"{name} e{ep}/{epochs}", leave=False)
        for x,y in it:
            x,y=x.to(device,non_blocking=True),y.to(device,non_blocking=True)
            with torch.no_grad(): t_out, t_feats = teacher(x, return_features=True)
            opt.zero_grad(set_to_none=True)
            s_out, s_feats = model(x, return_features=True)
            if method=="none":
                loss, ce_l, kd_l = ce(s_out,y), ce(s_out,y).detach(), torch.zeros((),device=device)
            elif adaptive:
                loss, ce_l, kd_l = adaptive_kd(s_out, t_out, y, T, lam, ce)
            else:
                loss, ce_l, kd_l = kd_loss(method, s_out, t_out, y, s_feats, t_feats, T, lam, ce)
            loss.backward(); opt.step()
            run+=loss.item()*y.size(0); ce_r+=ce_l.item()*y.size(0); kd_r+=float(kd_l)*y.size(0)
            corr+=s_out.argmax(1).eq(y).sum().item(); tot+=y.size(0)
        sch.step()
        vl_loss, vl_acc = evaluate(model, vl, ce)
        tr_acc=corr/tot
        for k,v in [("train_loss",run/tot),("train_acc",tr_acc),("val_loss",vl_loss),("val_acc",vl_acc),
                    ("ce",ce_r/tot),("kd",kd_r/tot),("lr",opt.param_groups[0]['lr']),("gap",tr_acc-vl_acc),("gconf",float('nan'))]:
            hist[k].append(v)
        if not np.isfinite(run): raise RuntimeError("NaN loss — diverged.")
        if vl_acc>best_val: best_val=vl_acc; best_state=copy.deepcopy(model.state_dict())
        if verbose: print(f"  {name} ep{ep:3d} train {tr_acc*100:5.2f}% val {vl_acc*100:5.2f}% ce {ce_r/tot:.3f} kd {kd_r/tot:.3f}")
    model.load_state_dict(best_state)
    vinfo = verify_ternary(model)
    row = dict(method=name, base_method=method, seed=seed, T=T, lam=lam, epochs=epochs,
               val_acc=best_val, sparsity=vinfo["sparsity"], all_ternary=vinfo["all_ternary"],
               final_train_acc=hist["train_acc"][-1])
    if record: REGISTRY.append(row)
    return dict(model=model, best_state=best_state, hist=hist, best_val=best_val, vinfo=vinfo, row=row)

def adaptive_kd(s, t, y, T, lam, ce):
    # teacher-confidence-weighted KD (Part 8): confident teacher -> stronger soft supervision.
    ce_l = ce(s, y)
    with torch.no_grad():
        pt = F.softmax(t, 1); ent = -(pt*torch.log(pt+1e-8)).sum(1)
        conf = 1 - ent/math.log(t.size(1)); conf = conf.clamp(0,1)      # 1=confident, 0=uniform
    logp = F.log_softmax(s/T,1); ptT = F.softmax(t/T,1)
    kl_per = (ptT*(torch.log(ptT+1e-8)-logp)).sum(1)*(T*T)
    kd = (conf*kl_per).mean()
    return (1-lam)*ce_l + lam*kd, ce_l.detach(), kd.detach()

def plot_curves(hist, title, save=None):
    ep=range(1,len(hist["train_acc"])+1)
    fig,ax=plt.subplots(1,2,figsize=(11,3.6))
    ax[0].plot(ep,[a*100 for a in hist["train_acc"]],label="train"); ax[0].plot(ep,[a*100 for a in hist["val_acc"]],label="val")
    ax[0].set_title(f"{title} — accuracy"); ax[0].set_xlabel("epoch"); ax[0].set_ylabel("%"); ax[0].legend(); ax[0].grid(alpha=.3)
    ax[1].plot(ep,hist["ce"],label="CE"); ax[1].plot(ep,hist["kd"],label="KD")
    ax[1].set_title(f"{title} — loss components"); ax[1].set_xlabel("epoch"); ax[1].legend(); ax[1].grid(alpha=.3)
    plt.tight_layout()
    if save: fig.savefig(save, dpi=120)
    plt.show()
print("training harness ready")
```

## Part 9 — Task 10: Gradient-Conflict Analysis

*Diagnostic.* Do the CE gradient and the KD gradient point in conflicting directions in the ternary
manifold? We measure `cos(∇CE, ∇KD)` over several batches for vanilla KD. Negative cosine ⇒ the two
objectives fight, which explains why naive KD can hurt and motivates loss balancing (only add balancing
if conflict is real).

```python
def grad_vector(model, loss):
    g = torch.autograd.grad(loss, [p for p in model.parameters() if p.requires_grad and p.ndim>=2],
                            retain_graph=True, allow_unused=True)
    return torch.cat([x.flatten() for x in g if x is not None])
set_seed(s0)
probe = build_student(QC, WARM_STATE)
ce = nn.CrossEntropyLoss(label_smoothing=CFG["LABEL_SMOOTHING"])
cosines=[]
it=iter(loaders[0])
for _ in range(20):
    try: x,y=next(it)
    except StopIteration: break
    x,y=x.to(device),y.to(device)
    with torch.no_grad(): t_out=teacher(x)
    s_out=probe(x)
    g_ce = grad_vector(probe, ce(s_out,y))
    g_kd = grad_vector(probe, _kl_T(s_out,t_out,CFG["KD_T"]))
    cosines.append(F.cosine_similarity(g_ce.unsqueeze(0),g_kd.unsqueeze(0)).item())
cosines=np.array(cosines)
fig,ax=plt.subplots(figsize=(7,3.5))
ax.hist(cosines,bins=15,color="#2b8cbe",edgecolor="k",alpha=.8); ax.axvline(0,color="red",ls="--")
ax.axvline(cosines.mean(),color="k",label=f"mean {cosines.mean():.3f}")
ax.set_xlabel("cos(∇CE, ∇KD)"); ax.set_title("Gradient conflict (vanilla KD, ternary student)"); ax.legend()
plt.tight_layout(); fig.savefig(out("plots","grad_conflict.png"),dpi=120); plt.show()
print(f"mean cos(gCE,gKD) = {cosines.mean():.3f}  ({'CONFLICT' if cosines.mean()<0 else 'aligned'})")
```

## Part 10 — Task 11: Consolidated Ablation Matrix

*Deliverable.* The assignment requires ≥1 ablation on T, λ, or Δ; this notebook produces several.
Here we run a compact **T×λ** grid at a short budget (the graded ablation) plus a quantizer ablation
(per-channel vs per-tensor, TWN vs TTQ). Reduce the grids if compute is tight.

```python
# --- T×λ ablation (short screen) ---
ABLA_EP = max(8, SE//3)
Ts=[1,2,4,8] if not QUICK_TEST else [2,4]; Ls=[0.3,0.5,0.7,0.9] if not QUICK_TEST else [0.5,0.9]
grid=np.zeros((len(Ts),len(Ls)))
for i,T in enumerate(Ts):
    for j,l in enumerate(Ls):
        r=train_student("vanilla", ABLA_EP, s0, loaders, teacher, QC, T=T, lam=l,
                        tag=f"abla_T{T}_l{l}", record=False, verbose=False)
        grid[i,j]=r["best_val"]; print(f"  T={T} λ={l} -> {r['best_val']*100:.2f}%")
fig,ax=plt.subplots(figsize=(1.3*len(Ls)+2,1.1*len(Ts)+2))
im=ax.imshow(grid*100,cmap="viridis",aspect="auto")
ax.set_xticks(range(len(Ls))); ax.set_xticklabels(Ls); ax.set_yticks(range(len(Ts))); ax.set_yticklabels(Ts)
ax.set_xlabel("λ"); ax.set_ylabel("T"); ax.set_title(f"T×λ ablation ({ABLA_EP}-ep screen)")
for i in range(len(Ts)):
    for j in range(len(Ls)): ax.text(j,i,f"{grid[i,j]*100:.1f}",ha="center",va="center",color="w",fontsize=8)
fig.colorbar(im,ax=ax,fraction=.046,pad=.04); plt.tight_layout()
fig.savefig(out("plots","ablation_T_lambda.png"),dpi=120); plt.show()
bi,bj=np.unravel_index(int(np.argmax(grid)),grid.shape)
print(f"Best (T,λ) = ({Ts[bi]}, {Ls[bj]}) -> {grid[bi,bj]*100:.2f}%")
CFG["KD_T"], CFG["KD_LAMBDA"] = Ts[bi], Ls[bj]
```

```python
# --- quantizer ablation ---
q_variants = {
  "per_channel_TWN_sym (default)": QuantConfig(scope="per_channel", threshold="twn", scale="derived", symmetric=True),
  "per_tensor_TWN_sym":            QuantConfig(scope="per_tensor",  threshold="twn", scale="derived", symmetric=True),
  "per_channel_TTQ_asym_learned":  QuantConfig(scope="per_channel", threshold="ttq", scale="learned", symmetric=False, t=0.05),
}
qabla={}
for name,qc in q_variants.items():
    r=train_student("none", SE, s0, loaders, teacher, qc, tag=f"qabla_{name}", record=False, verbose=False)
    qabla[name]=(r["best_val"], r["vinfo"]["sparsity"]); print(f"  {name}: {r['best_val']*100:.2f}% (sparsity {r['vinfo']['sparsity']*100:.1f}%)")
```

## Part 11 — Representation Analysis (why KD does / doesn't transfer)

We measure **linear CKA** between teacher and student penultimate features (how aligned the
representations are), feature norms, and a PCA of the student penultimate colored by class (separability).
This explains *which* teacher structure survives ternarization.

```python
def linear_cka(X, Y):
    X=X-X.mean(0,keepdim=True); Y=Y-Y.mean(0,keepdim=True)
    hsic=(X.t()@Y).pow(2).sum(); return (hsic/((X.t()@X).norm()*(Y.t()@Y).norm()+1e-8)).item()
@torch.no_grad()
def collect_penult(model, loader, n=2000):
    model.eval(); fs=[]; ys=[]; c=0
    for x,y in loader:
        x=x.to(device); _,f=model(x, return_features=True); fs.append(f["penultimate"].cpu()); ys.append(y); c+=y.size(0)
        if c>=n: break
    return torch.cat(fs)[:n], torch.cat(ys)[:n]

best_overall = train_student(BEST_METHOD, SE, s0, loaders, teacher, QC, tag=f"repr_{BEST_METHOD}", record=False, verbose=False)["model"]
Xs,ys = collect_penult(best_overall, loaders[1]); Xt,_ = collect_penult(teacher, loaders[1])
cka = linear_cka(Xs.to(device), Xt.to(device))
print(f"Linear CKA(student penultimate, teacher penultimate) = {cka:.3f}  (1=identical structure)")

from sklearn.decomposition import PCA
p2 = PCA(2).fit_transform(Xs.numpy())
fig,ax=plt.subplots(1,2,figsize=(11,4))
sc=ax[0].scatter(p2[:,0],p2[:,1],c=ys.numpy(),cmap="tab10",s=6,alpha=.6); ax[0].set_title(f"Student penultimate PCA (CKA vs teacher={cka:.2f})")
ax[1].bar(range(10),[Xs[ys==c].norm(dim=1).mean().item() for c in range(10)],color="#2b8cbe")
ax[1].set_xticks(range(10)); ax[1].set_xticklabels(CIFAR10_CLASSES,rotation=45,ha="right"); ax[1].set_title("Mean penultimate norm by class")
plt.tight_layout(); fig.savefig(out("plots","representation.png"),dpi=120); plt.show()
```

## Part 13 — Task 12: Compression Analysis + Full-Budget Confirmation

### 13a. Honest compression accounting

```python
def compression_report(model, arch_params_fp32):
    vinfo=verify_ternary(model)
    tern_w = sum(m.weight.numel() for m in model.modules() if isinstance(m,(TernaryConv2d,TernaryLinear)))
    nonzero = int(tern_w*(1-vinfo["sparsity"]))
    n_scales = sum(m.quant.g for m in model.modules() if isinstance(m,(TernaryConv2d,TernaryLinear)))
    fp32_bits = arch_params_fp32*32
    bits_2   = tern_w*2 + n_scales*32                    # 2-bit index + FP scales
    bits_158 = tern_w*math.log2(3) + n_scales*32         # entropy lower bound + scales
    return dict(ternary_weights=tern_w, nonzero=nonzero, sparsity=vinfo["sparsity"],
                fp32_MB=fp32_bits/8/1e6, ternary2bit_MB=bits_2/8/1e6, ternary158_MB=bits_158/8/1e6,
                ratio_2bit=fp32_bits/bits_2, ratio_158=fp32_bits/bits_158, all_ternary=vinfo["all_ternary"])
comp = compression_report(best_overall, n_params(resnet18()))
print(json.dumps({k:(round(v,3) if isinstance(v,float) else v) for k,v in comp.items()}, indent=2))
```

**Honesty (rubric-required).** The ~16–20× figure is a **storage** ratio (2-bit index, or the
log₂3≈1.58-bit entropy bound, plus FP32 per-channel scales). It is **not** a wall-clock speedup: PyTorch
stores latent FP32 weights and runs dense FP kernels, so measured inference time is unchanged. A real
speedup needs specialized ternary/bit-serial kernels or hardware. Report the storage ratio and this caveat.

```python
# 13b. Full-budget confirmation of the winning method (3 seeds)
print(f"Full-budget confirmation of BEST_METHOD='{BEST_METHOD}' at {CFG['FINAL_EPOCHS']} ep, seeds {CFG['SEEDS_FINAL']}, "
      f"(T={CFG['KD_T']}, λ={CFG['KD_LAMBDA']})")
final_runs=[]
for sd in CFG["SEEDS_FINAL"]:
    r=train_student(BEST_METHOD, CFG["FINAL_EPOCHS"], sd, loaders, teacher, QC,
                    T=CFG["KD_T"], lam=CFG["KD_LAMBDA"], tag=f"FINAL_{BEST_METHOD}_s{sd}", record=True, verbose=False)
    final_runs.append(r); print(f"  seed {sd}: val {r['best_val']*100:.2f}%  sparsity {r['vinfo']['sparsity']*100:.1f}%")
final_val=np.array([r["best_val"] for r in final_runs])
FINAL_MODEL = max(final_runs, key=lambda r:r["best_val"])["model"]
FINAL_STATE = max(final_runs, key=lambda r:r["best_val"])["best_state"]
torch.save({"model_state_dict":FINAL_STATE,"quant_config":vars(QC),"method":BEST_METHOD,
            "T":CFG["KD_T"],"lam":CFG["KD_LAMBDA"]}, out("checkpoints","final_ternary_student_best.pth"))
print(f"\nFINAL {BEST_METHOD}: {final_val.mean()*100:.2f}% ± {final_val.std()*100:.2f}  (best {final_val.max()*100:.2f}%)")
```

```python
# Compression-vs-accuracy Pareto scatter
fig,ax=plt.subplots(figsize=(7,5))
pts=[("teacher R34", teacher_val, 1.0),("FP32 R18", fp32_val, 1.0),
     ("ternary noKD", r_b2["best_val"], comp["ratio_158"]),
     (f"ternary {BEST_METHOD}", final_val.mean(), comp["ratio_158"])]
for name,acc,ratio in pts:
    ax.scatter(ratio, acc*100, s=80); ax.annotate(name,(ratio,acc*100),textcoords="offset points",xytext=(6,4),fontsize=9)
ax.set_xlabel("compression ratio vs FP32 (storage, ×)"); ax.set_ylabel("val accuracy (%)")
ax.set_title("Accuracy vs compression"); ax.grid(alpha=.3); plt.tight_layout()
fig.savefig(out("plots","pareto.png"),dpi=120); plt.show()
```

## Part 14 — Final Test Evaluation (firewalled, ONCE)

Only now do we unlock the official test set (C9/C11), on the single best final model. We also report
per-class accuracy, the confusion matrix, and calibration (ECE).

```python
test_loader = get_test_loader(CFG["BATCH_SIZE"], CFG["NUM_WORKERS"], CFG["DATA_ROOT"], final_evaluation=True)
@torch.no_grad()
def full_eval(model, loader):
    model.eval(); preds=[]; labels=[]; confs=[]
    for x,y in loader:
        x=x.to(device); p=F.softmax(model(x),1); c,pr=p.max(1)
        preds.append(pr.cpu()); labels.append(y); confs.append(c.cpu())
    return torch.cat(preds).numpy(), torch.cat(labels).numpy(), torch.cat(confs).numpy()
pr,lb,cf = full_eval(FINAL_MODEL, test_loader)
test_acc=(pr==lb).mean()
print(f"FINAL TERNARY STUDENT test accuracy: {test_acc*100:.2f}%")
_,t_test = evaluate(teacher, test_loader, ce_eval); print(f"Teacher test accuracy: {t_test*100:.2f}%")

# per-class + confusion + ECE
cm=np.zeros((10,10),int)
for p,t in zip(pr,lb): cm[t,p]+=1
fig,ax=plt.subplots(1,2,figsize=(13,5))
im=ax[0].imshow(cm,cmap="Blues"); ax[0].set_xticks(range(10)); ax[0].set_yticks(range(10))
ax[0].set_xticklabels(CIFAR10_CLASSES,rotation=45,ha="right"); ax[0].set_yticklabels(CIFAR10_CLASSES)
ax[0].set_title(f"Confusion (test acc {test_acc*100:.1f}%)"); ax[0].set_xlabel("pred"); ax[0].set_ylabel("true")
for i in range(10):
    for j in range(10): ax[0].text(j,i,cm[i,j],ha="center",va="center",fontsize=6,color="white" if cm[i,j]>cm.max()/2 else "black")
pca=[cm[i,i]/cm[i].sum()*100 for i in range(10)]
ax[1].bar(CIFAR10_CLASSES,pca,color="#2b8cbe"); ax[1].set_xticklabels(CIFAR10_CLASSES,rotation=45,ha="right")
ax[1].set_ylabel("accuracy %"); ax[1].set_title("Per-class test accuracy"); ax[1].set_ylim(0,100)
plt.tight_layout(); fig.savefig(out("plots","test_confusion.png"),dpi=120); plt.show()

# Expected Calibration Error
def ece(conf,pred,lab,bins=15):
    e=0
    for b in range(bins):
        lo,hi=b/bins,(b+1)/bins; m=(conf>lo)&(conf<=hi)
        if m.sum()>0: e+=abs((pred[m]==lab[m]).mean()-conf[m].mean())*m.mean()
    return e
print(f"Test ECE: {ece(cf,pr,lb):.4f}")
```

   hypothesis: *transform teacher knowledge into an attainable target rather than forcing imitation.*
2. **Where did the accuracy budget go?** capacity gap vs quantization gap vs KD recovery — report all three.
3. **Compression:** report the storage ratio **and** the honest no-real-speedup-without-kernels caveat.
4. **Ablations:** the T×λ heatmap (graded), the STE study, progressive test, quantizer ablation, and the
   gradient-conflict histogram are all saved under `results_tasks5_12/plots/`.
5. **Limitations:** ternary impact concentrates on high-quant-error layers (Part 12); first-conv/FC remain
   ternary by constraint; results are single-GPU, few-seed at the screening budget — the final method is
   the only 3-seed number.
