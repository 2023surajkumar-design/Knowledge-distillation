# ATDL final-two — validated post-Task-4 continuation

This notebook is artifact-first: Tasks 1–4 are frozen and never retrained here. It verifies their real paths and split lineage, then runs one isolated QTRD validation screen with fresh output names. The official test set remains locked and is not accessed.

```python
from pathlib import Path
import json

ROOT = Path('/home/vu-lab03-pc17/ATDL-1').resolve()
DATA_ROOT = ROOT / 'data'
STATE_PATH = ROOT / 'research_state_checkpoint.json'
TASK1_TEACHER = ROOT / 'resnet34_cifar10_fp32_best.pth'
TASK2_STUDENT = ROOT / 'results/best_models/task2_b1_fp32_resnet18_best.pth'
TASK3_TERNARY = ROOT / 'results/best_models/task3_b2_default_best.pth'
TASK3_SUMMARY = ROOT / 'results/task3_b2_default_summary.json'
TASK4_AGGREGATE = ROOT / 'results/task4/task4_b3_final_t2_lam09_aggregate_summary.json'
TASK4_BEST = ROOT / 'experiments/task4/checkpoints/task4_b3_final_t2_lam09_remaining_rerun1/resnet18_ternary_kd_seed44.pth'
QTRD_CONFIG = ROOT / 'configs/kd/resnet18_ternary_qtrd_finaltwo_screen.yaml'
QTRD_PIPELINE = ROOT / 'scripts/run_final_two_notebook_pipeline.py'

required = [DATA_ROOT, STATE_PATH, TASK1_TEACHER, TASK2_STUDENT, TASK3_TERNARY, TASK3_SUMMARY, TASK4_AGGREGATE, TASK4_BEST, QTRD_CONFIG, QTRD_PIPELINE]
missing = [str(path) for path in required if not path.exists()]
assert not missing, f'Missing required artifact(s): {missing}'
print('Validated ATDL root and all frozen Task 1–4 / QTRD paths.')
```

```python
task4 = json.loads(TASK4_AGGREGATE.read_text())
task3 = json.loads(TASK3_SUMMARY.read_text())
assert task4['test_evaluation'] == 'not_run'
assert task3['test_evaluation'] == 'not_run'
assert task4['validation_mean'] == 0.9512
per_seed = {str(row['seed']): row['best_validation_accuracy'] for row in task4['per_seed']}
assert per_seed == {'42': 0.9514, '43': 0.9496, '44': 0.9526}
print('Frozen selection: Task 4 vanilla KD, 95.12% ± 0.151% validation (seeds 42/43/44).')
print('Split lineage is inherited from the production trainer; no notebook-local split or model implementation is allowed.')
```
