# Task 1 — Training ResNet34 from Scratch on CIFAR-10

**Full-precision (FP32) ResNet34 teacher for a later Knowledge-Distillation + Ternary-Quantization experiment.**

---

## 1. Objective

The single goal of this notebook is to train a **strong, well-generalized, full-precision (FP32) ResNet34** on CIFAR-10 **from scratch**, and to save a clean, reproducible checkpoint.

That checkpoint will later be **frozen** and used as the **teacher** in the pipeline:

```
FP32 ResNet34 teacher  ->  knowledge distillation  ->  ternary ResNet18 student
```

**This notebook is ONLY Task 1.** It deliberately contains **no** knowledge distillation, **no** ternarization, **no** quantization-aware training, **no** student model, and **no** feature/KD/quantization losses. Those belong to Task 2/3.

**Hard constraints honored here**
- Architecture: **ResNet34** with the canonical `[3, 4, 6, 3]` `BasicBlock` configuration.
- Precision: **FP32** throughout.
- Trained **entirely from scratch** — no ImageNet weights, no pretrained checkpoint, no external data.
- Dataset: **CIFAR-10** only; the official test set is never used as a training source.

**What "good" means here.** The objective is **high test accuracy with stable convergence and a reasonable train/test gap** — *not* minimizing the train/test gap for its own sake, and *not* maximizing training accuracy. A model can have a tiny gap simply by being weak (low train *and* low test accuracy). We therefore report train accuracy, test accuracy, and the gap **together**.

> **On reported numbers.** This configuration is designed to be a strong CIFAR-10 ResNet34 baseline. Well-tuned setups of this kind commonly land in roughly the **94–95% test-accuracy** range, but the exact number you obtain depends on hardware, PyTorch/torchvision versions, the random seed, and how long you train. We do **not** hard-code or promise a specific accuracy.

## 2. Environment and Reproducibility

We set seeds for Python, NumPy, and PyTorch (CPU + CUDA) so runs are repeatable. This training run requires an available CUDA GPU; it never silently falls back to CPU.

**Reproducibility vs. speed — the trade-off.** Forcing fully deterministic cuDNN kernels (`torch.backends.cudnn.deterministic = True`, `benchmark = False`, and `torch.use_deterministic_algorithms(True)`) removes most run-to-run variation, but some deterministic convolution kernels are **substantially slower**, and a few ops have no deterministic implementation at all (they would raise an error). For training a CNN for ~200 epochs, that slowdown is costly.

Our default is a **pragmatic middle ground**: we seed everything (so the data order, augmentation, and initialization are fixed) but leave `cudnn.benchmark = True` for speed. This makes results *close to* reproducible and fast. Flip `STRICT_DETERMINISM = True` in the config if you need bit-exact reproducibility and can accept the speed hit.

## 6. Visualize the Dataset

Looking at raw images before training is a cheap, high-value sanity check: it confirms the data loaded correctly, labels line up with images, and there is nothing obviously corrupted or mis-decoded. It also builds intuition for how small and low-resolution CIFAR-10 really is (32×32), which is exactly why we adapt the ResNet stem later.

````python
def show_image_grid(dataset, class_names, n=16, ncols=8, title="CIFAR-10 samples"):
    # Display a grid of images with their class labels. Expects tensors in [0,1].
    n = min(n, len(dataset))
    nrows = math.ceil(n / ncols)
    idxs = random.sample(range(len(dataset)), n)
    fig, axes = plt.subplots(nrows, ncols, figsize=(1.6 * ncols, 1.9 * nrows))
    axes = np.array(axes).reshape(-1)
    for ax in axes:
        ax.axis("off")
    for ax_i, ds_i in enumerate(idxs):
        img, lbl = dataset[ds_i]
        img = img.permute(1, 2, 0).numpy()   # C,H,W -> H,W,C
        img = np.clip(img, 0, 1)
        axes[ax_i].imshow(img)
        axes[ax_i].set_title(class_names[lbl], fontsize=9)
    fig.suptitle(title, fontsize=13)
    fig.tight_layout()
    plt.show()

show_image_grid(_raw_train, CIFAR10_CLASSES, n=16, ncols=8,
                title="Random CIFAR-10 training images (raw)")
````

## 7. Data Preprocessing and Augmentation

We design preprocessing **for CIFAR-10 specifically** — not the ImageNet 224×224 pipeline.

**Training augmentation (baseline):**
- `RandomCrop(32, padding=4)` — pad 4 pixels on each side (reflection-free zero pad) then crop back to 32×32. This simulates small **translations**, teaching the network that an object's identity is invariant to where it sits in the frame.
- `RandomHorizontalFlip()` — left/right mirror with p=0.5. Most CIFAR-10 classes are horizontally symmetric in identity (a mirrored dog is still a dog), so this roughly doubles effective data for free.

Both augmentations attack **overfitting**: they enlarge the effective training distribution so the model must learn robust, generalizable features rather than memorizing pixel-exact training images.

**Normalization.** We standardize each channel to zero mean / unit variance using the well-established CIFAR-10 statistics. Centering/scaling inputs stabilizes and speeds up optimization. If `COMPUTE_NORM_STATS = True`, we instead compute mean/std **from the training split only** — never from the test set, which would leak test information into preprocessing.

**Test transform.** The test pipeline is `ToTensor + Normalize` **only** — no random crop, no flip. Evaluation must be **deterministic** and reflect performance on the original images; random test-time augmentation would make the reported accuracy noisy and non-comparable across runs (that is a separate technique, TTA, not part of a standard baseline evaluation).

````python
# Standard, widely-used CIFAR-10 channel statistics.
CIFAR10_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR10_STD  = (0.2470, 0.2435, 0.2616)

def compute_train_mean_std(data_root):
    # Compute per-channel mean/std from the TRAINING split only.
    ds = torchvision.datasets.CIFAR10(root=data_root, train=True, download=True,
                                      transform=transforms.ToTensor())
    loader = DataLoader(ds, batch_size=512, shuffle=False, num_workers=0)
    n_pixels = 0
    channel_sum = torch.zeros(3)
    channel_sq_sum = torch.zeros(3)
    for imgs, _ in loader:
        # imgs: (B, 3, 32, 32)
        b, c, h, w = imgs.shape
        n_pixels += b * h * w
        channel_sum += imgs.sum(dim=[0, 2, 3])
        channel_sq_sum += (imgs ** 2).sum(dim=[0, 2, 3])
    mean = channel_sum / n_pixels
    std = torch.sqrt(channel_sq_sum / n_pixels - mean ** 2)
    return tuple(mean.tolist()), tuple(std.tolist())

if CONFIG["COMPUTE_NORM_STATS"]:
    MEAN, STD = compute_train_mean_std(CONFIG["DATA_ROOT"])
    print("Computed from TRAIN split:")
else:
    MEAN, STD = CIFAR10_MEAN, CIFAR10_STD
    print("Using standard constants:")
print(f"  mean = {tuple(round(m, 4) for m in MEAN)}")
print(f"  std  = {tuple(round(s, 4) for s in STD)}")

train_transform = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

test_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

print("\nTrain transform:\n", train_transform)
print("\nTest transform:\n", test_transform)
````

## 8. DataLoaders

We now build the datasets **with the correct transforms** and wrap them in `DataLoader`s.

- **Batch size** controls the gradient-noise/throughput trade-off. Larger batches give smoother gradient estimates and better GPU utilization but may need LR re-tuning (handled by the linear-scaling rule in the config) and can slightly hurt generalization if pushed too far.
- **`num_workers`** parallelizes CPU-side data loading/augmentation so the GPU is not starved. More workers help on Linux/Colab; on Windows notebooks `0` is the safe default (subprocess spawning can hang).
- **`pin_memory=True`** (enabled only on CUDA) uses page-locked host memory, making the host→GPU copy faster.

The **training** loader shuffles every epoch (decorrelates mini-batches, improves SGD); the **test** loader does not shuffle (order is irrelevant and non-shuffling keeps evaluation reproducible).

````python
# Build a fixed, class-balanced 45k/5k split from the official training set.
# The validation view has no random augmentation; the official test set remains final-evaluation-only.
train_dataset_base = torchvision.datasets.CIFAR10(
    root=CONFIG["DATA_ROOT"], train=True, download=True, transform=train_transform)
validation_dataset_base = torchvision.datasets.CIFAR10(
    root=CONFIG["DATA_ROOT"], train=True, download=True, transform=test_transform)
test_dataset = torchvision.datasets.CIFAR10(
    root=CONFIG["DATA_ROOT"], train=False, download=True, transform=test_transform)

val_per_class = int(len(train_dataset_base) * CONFIG["VALIDATION_FRACTION"] / len(CIFAR10_CLASSES))
if val_per_class <= 0:
    raise ValueError("VALIDATION_FRACTION is too small to allocate one sample per class.")
split_generator = torch.Generator().manual_seed(CONFIG["SEED"])
targets = np.asarray(train_dataset_base.targets)
train_indices, val_indices = [], []
for class_idx in range(len(CIFAR10_CLASSES)):
    class_indices = np.flatnonzero(targets == class_idx)
    permuted = class_indices[torch.randperm(len(class_indices), generator=split_generator).numpy()]
    val_indices.extend(permuted[:val_per_class].tolist())
    train_indices.extend(permuted[val_per_class:].tolist())

train_dataset = Subset(train_dataset_base, train_indices)
validation_dataset = Subset(validation_dataset_base, val_indices)
assert len(train_dataset) + len(validation_dataset) == 50000
assert set(train_indices).isdisjoint(val_indices)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    random.seed(worker_seed)
    np.random.seed(worker_seed)

_use_pin = torch.cuda.is_available()
_nw = CONFIG["NUM_WORKERS"]
_persistent = _nw > 0
loader_generator = torch.Generator().manual_seed(CONFIG["SEED"])
loader_common = dict(
    num_workers=_nw, pin_memory=_use_pin, persistent_workers=_persistent,
    worker_init_fn=seed_worker, generator=loader_generator,
)

train_loader = DataLoader(train_dataset, batch_size=CONFIG["BATCH_SIZE"], shuffle=True,
                          drop_last=False, **loader_common)
validation_loader = DataLoader(validation_dataset, batch_size=CONFIG["BATCH_SIZE"], shuffle=False,
                               drop_last=False, **loader_common)
test_loader = DataLoader(test_dataset, batch_size=CONFIG["BATCH_SIZE"], shuffle=False,
                         drop_last=False, **loader_common)

print(f"Batch size           : {CONFIG['BATCH_SIZE']}")
print(f"num_workers          : {_nw}")
print(f"pin_memory           : {_use_pin}")
print(f"Train/validation     : {len(train_dataset):,} / {len(validation_dataset):,} (stratified)")
print(f"Official test set    : {len(test_dataset):,} (final evaluation only)")
print(f"Train/val/test batches: {len(train_loader)} / {len(validation_loader)} / {len(test_loader)}")

_xb, _yb = next(iter(train_loader))
print(f"One train batch: images {tuple(_xb.shape)}, labels {tuple(_yb.shape)}")
assert _xb.shape[1:] == (3, 32, 32)
````

## 9. CIFAR-10 ResNet34 Architecture

We define ResNet34 **explicitly** (rather than loading `torchvision.models.resnet34`) so the architecture is transparent, randomly initialized, and correctly adapted for 32×32 inputs.

**Canonical ImageNet ResNet34 vs. CIFAR-10-adapted ResNet34**

| Component | ImageNet ResNet34 | CIFAR-10 adaptation (this notebook) |
|---|---|---|
| Stem conv | 7×7, stride 2 | **3×3, stride 1, padding 1** |
| Stem pool | 3×3 max-pool, stride 2 | **removed (Identity)** |
| Input size | 224×224 | 32×32 |
| Block config | `[3, 4, 6, 3]` BasicBlock | **`[3, 4, 6, 3]` BasicBlock (unchanged)** |
| Stage channels | 64, 128, 256, 512 | 64, 128, 256, 512 (unchanged) |
| Head | GAP → FC(1000) | GAP → **FC(10)** |

**Why change the stem?** ImageNet's 7×7 stride-2 conv followed by a stride-2 max-pool downsamples the input by 4× immediately — sensible for 224×224 images, but catastrophic for 32×32 CIFAR images, where it would shrink the feature map to 8×8 before any real learning happens, discarding most spatial detail. Replacing the stem with a **3×3 stride-1 conv and no max-pool** preserves the 32×32 resolution through the early layers, which is the standard, well-established CIFAR ResNet recipe.

**Crucially, the model remains recognizably ResNet34**: the same `BasicBlock` (two 3×3 convs + identity/projection shortcut), the same `[3, 4, 6, 3]` stage depths, the same channel widths, and the same downsampling-by-stride-2 at the start of stages 2–4. Only the stem and the final class count change.

````python
class BasicBlock(nn.Module):
    # Standard ResNet BasicBlock: two 3x3 convs, BN+ReLU, identity/projection shortcut.
    expansion = 1

    def __init__(self, in_channels, out_channels, stride=1):
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3,
                               stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3,
                               stride=1, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(out_channels)

        # Shortcut: identity when shapes match, else a 1x1 projection conv to match dims/stride.
        self.shortcut = nn.Sequential()
        if stride != 1 or in_channels != out_channels * self.expansion:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_channels, out_channels * self.expansion, kernel_size=1,
                          stride=stride, bias=False),
                nn.BatchNorm2d(out_channels * self.expansion),
            )

    def forward(self, x):
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out = out + self.shortcut(x)   # residual connection
        out = F.relu(out)
        return out


class ResNetCIFAR(nn.Module):
    # ResNet for CIFAR-10 with a 3x3 stride-1 stem and no initial max-pool.
    def __init__(self, block, num_blocks, num_classes=10):
        super().__init__()
        self.in_channels = 64

        # CIFAR stem: 3x3 stride-1, padding 1, NO max-pool (preserves 32x32).
        self.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(64)

        # Four stages with the ResNet34 depths [3, 4, 6, 3].
        self.layer1 = self._make_layer(block, 64,  num_blocks[0], stride=1)
        self.layer2 = self._make_layer(block, 128, num_blocks[1], stride=2)
        self.layer3 = self._make_layer(block, 256, num_blocks[2], stride=2)
        self.layer4 = self._make_layer(block, 512, num_blocks[3], stride=2)

        self.avgpool = nn.AdaptiveAvgPool2d((1, 1))   # global average pooling
        self.fc = nn.Linear(512 * block.expansion, num_classes)

        self._initialize_weights()

    def _make_layer(self, block, out_channels, num_blocks, stride):
        # First block in a stage may downsample (stride); the rest use stride 1.
        strides = [stride] + [1] * (num_blocks - 1)
        layers = []
        for s in strides:
            layers.append(block(self.in_channels, out_channels, s))
            self.in_channels = out_channels * block.expansion
        return nn.Sequential(*layers)

    def _initialize_weights(self):
        # Kaiming init for convs; standard init for BN and the linear head.
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode="fan_out", nonlinearity="relu")
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.normal_(m.weight, 0, 0.01)
                nn.init.constant_(m.bias, 0)

    def forward(self, x):
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.layer1(out)
        out = self.layer2(out)
        out = self.layer3(out)
        out = self.layer4(out)
        out = self.avgpool(out)
        out = torch.flatten(out, 1)
        out = self.fc(out)
        return out


def ResNet34(num_classes=10):
    # ResNet34 = BasicBlock with stage depths [3, 4, 6, 3].
    return ResNetCIFAR(BasicBlock, [3, 4, 6, 3], num_classes=num_classes)

print("ResNet34 (CIFAR-adapted) defined.")
````

## 10. Model Sanity Checks

Before training we verify the model is well-formed: correct output shape, FP32 parameters, a sane parameter count (a ResNet34 with a 10-class head has ~21.3M parameters), and — importantly — that **no pretrained weights** were loaded (we only ever call `ResNet34()`, which random-initializes).

````python
model = ResNet34(num_classes=10).to(device)

# Parameter accounting.
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
param_bytes = sum(p.numel() * p.element_size() for p in model.parameters())

print(model)
print("\n==== Model Summary ====")
print(f"Total parameters      : {total_params:,}")
print(f"Trainable parameters  : {trainable_params:,}")
print(f"Approx FP32 param mem : {param_bytes / (1024**2):.2f} MB")

# Forward a dummy batch and check the output shape.
model.eval()
with torch.no_grad():
    dummy = torch.randn(8, 3, 32, 32, device=device)
    out = model(dummy)
print(f"Dummy input shape     : {tuple(dummy.shape)}")
print(f"Dummy output shape    : {tuple(out.shape)}")

# Assertions.
assert out.shape == (8, 10), f"Output must be [batch, 10], got {tuple(out.shape)}"
assert all(p.dtype == torch.float32 for p in model.parameters()), "All params must be FP32"
assert model.fc.out_features == 10, "Classifier must output exactly 10 classes"
print("\nAll sanity checks passed: output=[B,10], FP32 params, 10-class head, random init (no pretrained weights).")
````

## 11. Loss Function

We use **cross-entropy**, the standard loss for single-label multi-class classification, optionally with **label smoothing**.

- **Standard cross-entropy** pushes the predicted probability of the correct class toward 1 and all others toward 0. With very confident models this can encourage over-confident, poorly-calibrated outputs.
- **Label-smoothed cross-entropy** replaces the hard one-hot target with a soft target that puts `1 - ε` on the true class and spreads `ε` uniformly over the rest (here `ε = LABEL_SMOOTHING`). This discourages over-confidence, often improves calibration, and can modestly improve test accuracy.

There is **no KD loss, no feature loss, no ternary/quantization loss** here — Task 1 is a plain supervised baseline.

````python
criterion = nn.CrossEntropyLoss(label_smoothing=CONFIG["LABEL_SMOOTHING"])
print("Loss:", criterion)
print(f"Label smoothing: {CONFIG['LABEL_SMOOTHING']}")
````

## 12. Optimizer and Scheduler

**Optimizer — SGD with Nesterov momentum.** SGD+momentum remains the strongest, most reliable optimizer for training CNNs on CIFAR-10 from scratch: it generalizes better than adaptive methods (Adam/RMSProp) in this regime and interacts cleanly with weight decay and cosine LR schedules. Momentum (0.9) accelerates progress along consistent gradient directions and damps oscillations.

**Weight decay (5e-4)** is an L2 penalty on the weights that shrinks them toward zero, discouraging overly complex functions and reducing overfitting — this typically **narrows the train/test gap while improving test accuracy**. Note: weight decay is *not* the same as dropout (dropout randomly zeros activations; weight decay penalizes weight magnitude). We use weight decay and no dropout, which is standard for ResNets.

**Scheduler — linear warmup + cosine annealing.** The LR starts low, ramps linearly for `WARMUP_EPOCHS` (stabilizing the first, high-variance updates), then follows a cosine curve down to ~0. Cosine decay spends more time at moderate LRs and eases into a small final LR, which tends to find flatter, better-generalizing minima than abrupt step drops. We implement it with `LambdaLR` (a multiplier on the base LR) so it is robust across PyTorch versions, and we **step it once per epoch**.

````python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=CONFIG["EFFECTIVE_LR"],
    momentum=CONFIG["MOMENTUM"],
    weight_decay=CONFIG["WEIGHT_DECAY"],
    nesterov=CONFIG["NESTEROV"],
)

def make_lr_lambda(warmup_epochs, total_epochs):
    def lr_lambda(epoch):
        # epoch is 0-indexed by LambdaLR.
        if warmup_epochs > 0 and epoch < warmup_epochs:
            return float(epoch + 1) / float(warmup_epochs)
        # Cosine decay over the remaining epochs, from multiplier 1 -> 0.
        progress = (epoch - warmup_epochs) / max(1, (total_epochs - warmup_epochs))
        progress = min(max(progress, 0.0), 1.0)
        return 0.5 * (1.0 + math.cos(math.pi * progress))
    return lr_lambda

scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=make_lr_lambda(CONFIG["WARMUP_EPOCHS"], CONFIG["EPOCHS"]))

print("Optimizer:", optimizer)
print(f"Effective base LR: {CONFIG['EFFECTIVE_LR']:.5f} "
      f"(base {CONFIG['LEARNING_RATE']} scaled for batch size {CONFIG['BATCH_SIZE']})")
````

````python
# Preview the LR schedule WITHOUT disturbing the real optimizer/scheduler state.
def preview_lr_schedule(base_lr, warmup_epochs, total_epochs):
    fn = make_lr_lambda(warmup_epochs, total_epochs)
    return [base_lr * fn(e) for e in range(total_epochs)]

_lr_curve = preview_lr_schedule(CONFIG["EFFECTIVE_LR"], CONFIG["WARMUP_EPOCHS"], CONFIG["EPOCHS"])

plt.figure(figsize=(7, 4))
plt.plot(range(1, CONFIG["EPOCHS"] + 1), _lr_curve)
plt.xlabel("Epoch")
plt.ylabel("Learning rate")
plt.title("Planned LR schedule (linear warmup + cosine decay)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print(f"LR start (epoch 1): {_lr_curve[0]:.5f}")
print(f"LR peak           : {max(_lr_curve):.5f}")
print(f"LR end            : {_lr_curve[-1]:.6f}")
````

## 13. Training Function

`train_one_epoch` runs a single pass over the training data and returns the **average training loss** and **training accuracy**. It puts the model in `train()` mode (so BatchNorm updates its running statistics and uses batch statistics), moves data to the device, and performs the standard forward → loss → backward → step cycle. We use `optimizer.zero_grad(set_to_none=True)` (slightly faster and more memory-friendly than zeroing).

````python
def train_one_epoch(model, loader, criterion, optimizer, device, epoch=None, epochs=None):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    desc = f"Train {epoch}/{epochs}" if epoch is not None else "Train"
    pbar = tqdm(loader, desc=desc, leave=False)
    for inputs, labels in pbar:
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        correct += predicted.eq(labels).sum().item()
        total += labels.size(0)

        # Live batch metrics on the progress bar (guarded for the no-tqdm fallback).
        if hasattr(pbar, "set_postfix"):
            pbar.set_postfix(loss=running_loss / total, acc=100.0 * correct / total)

    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc
````

## 14. Evaluation Function

`evaluate` measures loss and accuracy on a loader with **no gradient tracking** (`torch.no_grad()`) and the model in `eval()` mode. `eval()` is essential: it makes BatchNorm use its **running (population) statistics** instead of batch statistics, so evaluation is deterministic and does not depend on batch composition. No augmentation is applied here — the test loader uses the deterministic `test_transform`. The function can optionally return all predictions/labels, which we reuse later for the confusion matrix and per-class accuracy.

````python
@torch.no_grad()
def evaluate(model, loader, criterion, device, return_preds=False):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    all_preds = []
    all_labels = []

    for inputs, labels in loader:
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        outputs = model(inputs)
        loss = criterion(outputs, labels)

        running_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        correct += predicted.eq(labels).sum().item()
        total += labels.size(0)

        if return_preds:
            all_preds.append(predicted.cpu())
            all_labels.append(labels.cpu())

    loss_val = running_loss / total
    acc_val = correct / total
    if return_preds:
        return loss_val, acc_val, torch.cat(all_preds).numpy(), torch.cat(all_labels).numpy()
    return loss_val, acc_val
````

## 15 & 16. Full Training Loop + Checkpointing

We now train for `EPOCHS`, recording train and held-out-validation loss/accuracy plus the learning rate each epoch. We **step the scheduler once per epoch** (after the optimizer steps for that epoch).

**Validation protocol.** A fixed, class-balanced 5,000-image validation partition is carved from the official training split. It is used for per-epoch monitoring and checkpoint selection. The official 10,000-image CIFAR-10 test set is evaluated only after the selected checkpoint is frozen, avoiding repeated-test selection bias.

**Checkpointing.** Whenever validation accuracy improves, we save a rich checkpoint: model/optimizer/scheduler state, epoch, best accuracy, the config, and the seed — enough to resume or reproduce.

````python
def save_checkpoint(path, model, optimizer, scheduler, epoch, best_acc, config, extra=None):
    payload = {
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": scheduler.state_dict(),
        "epoch": epoch,
        "best_accuracy": best_acc,
        "config": config,
        "seed": config["SEED"],
        "arch": "ResNet34-CIFAR",
        "num_classes": 10,
        "normalization": {"mean": list(MEAN), "std": list(STD)},
    }
    if extra:
        payload.update(extra)
    torch.save(payload, path)


history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": [],
           "lr": [], "gap": []}
best_acc = 0.0
best_epoch = -1

print(f"Training ResNet34 for {CONFIG['EPOCHS']} epochs on {device} ...")
_start = time.time()

for epoch in range(1, CONFIG["EPOCHS"] + 1):
    current_lr = optimizer.param_groups[0]["lr"]

    train_loss, train_acc = train_one_epoch(
        model, train_loader, criterion, optimizer, device, epoch, CONFIG["EPOCHS"])
    val_loss, val_acc = evaluate(model, validation_loader, criterion, device)

    scheduler.step()   # step once per epoch

    gap = train_acc - val_acc
    history["train_loss"].append(train_loss)
    history["train_acc"].append(train_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    history["lr"].append(current_lr)
    history["gap"].append(gap)

    improved = val_acc > best_acc
    if improved:
        best_acc = val_acc
        best_epoch = epoch
        save_checkpoint(CONFIG["CHECKPOINT_PATH"], model, optimizer, scheduler,
                        epoch, best_acc, CONFIG)

    print(f"Epoch {epoch:3d}/{CONFIG['EPOCHS']} | "
          f"Train Loss {train_loss:.4f} Acc {100*train_acc:5.2f}% | "
          f"Val Loss {val_loss:.4f} Acc {100*val_acc:5.2f}% | "
          f"Gap {100*gap:5.2f}% | LR {current_lr:.5f}"
          + ("  <-- best" if improved else ""))

_elapsed = time.time() - _start
print(f"\nTraining complete in {_elapsed/60:.1f} min.")
print(f"Best validation accuracy: {100*best_acc:.2f}% at epoch {best_epoch}")

# Persist history for later analysis / the KD experiment.
with open(CONFIG["HISTORY_PATH"], "w") as f:
    json.dump({"history": history, "best_val_acc": best_acc, "best_epoch": best_epoch,
               "config": CONFIG, "elapsed_sec": _elapsed}, f, indent=2)
print(f"Saved history to {CONFIG['HISTORY_PATH']}")
````

## 17. Training Curves

Four diagnostic figures. What to look for:

- **Underfitting** — both train and test loss stay high / accuracy stays low: the model or training budget is too small, or the LR is off.
- **Overfitting** — train accuracy keeps rising while test accuracy stalls or falls, and the gap widens: needs more regularization/augmentation or fewer epochs.
- **Healthy convergence** — both curves rise together, test accuracy plateaus high, and the gap is modest and stable.
- **Unstable optimization** — loss spikes or oscillates wildly: usually the LR is too high (or warmup too short).

````python
_epochs_axis = range(1, len(history["train_loss"]) + 1)

# FIGURE 1: Loss.
plt.figure(figsize=(7, 4.5))
plt.plot(_epochs_axis, history["train_loss"], label="Train loss")
plt.plot(_epochs_axis, history["val_loss"], label="Validation loss")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Figure 1 — Training vs. Validation Loss")
plt.legend(); plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

# FIGURE 2: Accuracy.
plt.figure(figsize=(7, 4.5))
plt.plot(_epochs_axis, [100*a for a in history["train_acc"]], label="Train accuracy")
plt.plot(_epochs_axis, [100*a for a in history["val_acc"]], label="Validation accuracy")
plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)"); plt.title("Figure 2 — Training vs. Validation Accuracy")
plt.legend(); plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

# FIGURE 3: Generalization gap.
plt.figure(figsize=(7, 4.5))
plt.plot(_epochs_axis, [100*g for g in history["gap"]], color="tab:red",
         label="Train acc - Validation acc")
plt.axhline(0, color="k", linewidth=0.8, alpha=0.5)
plt.xlabel("Epoch"); plt.ylabel("Gap (percentage points)")
plt.title("Figure 3 — Generalization Gap vs. Epoch")
plt.legend(); plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

# FIGURE 4: Learning rate.
plt.figure(figsize=(7, 4.5))
plt.plot(_epochs_axis, history["lr"], color="tab:green")
plt.xlabel("Epoch"); plt.ylabel("Learning rate"); plt.title("Figure 4 — Learning Rate vs. Epoch")
plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()
````

## 18. Generalization Gap Analysis

A small gap alone is **not** the goal. The table below pairs the gap with the actual test accuracy so the two are always read together. A healthy result is **high test accuracy with a modest, stable gap**; a small gap with mediocre test accuracy just means the model is uniformly weak.

````python
if len(history["val_acc"]) > 0:
    final_train = history["train_acc"][-1]
    final_val = history["val_acc"][-1]
    final_gap = final_train - final_val
    peak_val = max(history["val_acc"])
    peak_val_epoch = int(np.argmax(history["val_acc"])) + 1

    print("==== Generalization Summary ====")
    print(f"Final train accuracy : {100*final_train:.2f}%")
    print(f"Final validation accuracy  : {100*final_val:.2f}%")
    print(f"Final train-validation gap : {100*final_gap:.2f} pp")
    print(f"Best validation accuracy   : {100*peak_val:.2f}% (epoch {peak_val_epoch})")
    print()
    if final_gap > 0.08 and final_val < peak_val - 0.005:
        print("Interpretation: sizable gap + validation below its peak => some overfitting; "
              "consider stronger augmentation, more weight decay, or early stopping at the best epoch.")
    elif peak_val < 0.85:
        print("Interpretation: validation accuracy is modest => likely under-trained "
              "(increase EPOCHS) or under-tuned (revisit LR); the small gap is not by itself good news.")
    else:
        print("Interpretation: strong validation accuracy with a controlled gap => healthy generalization.")
else:
    print("No history recorded (did training run?).")
````

## 21. Confusion Matrix

The confusion matrix shows *which* classes get mistaken for *which* others. Strong off-diagonal cells (e.g., cat↔dog) indicate systematic confusion. We show both raw counts and a row-normalized version (each row sums to 1, i.e., per-true-class distribution). Uses scikit-learn if available, otherwise a manual implementation.

````python
def manual_confusion_matrix(y_true, y_pred, n):
    cm = np.zeros((n, n), dtype=int)
    for t, p in zip(y_true, y_pred):
        cm[t, p] += 1
    return cm

if _HAS_SKLEARN:
    cm = sk_confusion_matrix(all_labels, all_preds, labels=list(range(num_classes)))
else:
    cm = manual_confusion_matrix(all_labels, all_preds, num_classes)

cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

def plot_cm(matrix, title, fmt):
    fig, ax = plt.subplots(figsize=(7.5, 6.5))
    im = ax.imshow(matrix, cmap="Blues")
    ax.set_xticks(range(num_classes)); ax.set_yticks(range(num_classes))
    ax.set_xticklabels(CIFAR10_CLASSES, rotation=45, ha="right")
    ax.set_yticklabels(CIFAR10_CLASSES)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(title)
    thresh = matrix.max() / 2.0
    for i in range(num_classes):
        for j in range(num_classes):
            ax.text(j, i, format(matrix[i, j], fmt), ha="center", va="center",
                    color="white" if matrix[i, j] > thresh else "black", fontsize=8)
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    fig.tight_layout(); plt.show()

plot_cm(cm, "Confusion Matrix (counts)", "d")
plot_cm(cm_norm, "Confusion Matrix (row-normalized)", ".2f")
````

## 25. Hyperparameter Tuning Guide

Tune **one thing at a time** and judge every change by **test (or validation) accuracy plus the gap**, not training accuracy. A sensible order:

1. **Learning rate** — the single highest-impact knob. If training is unstable/diverges, lower it (or lengthen warmup); if it plateaus early and underfits, raise it. With batch size 128, 0.1 is a strong start.
2. **Weight decay** — if the gap is large and test accuracy trails train, increase it (e.g., 5e-4 → 1e-3); if the model underfits, decrease it. This most directly trades off the train/test gap.
3. **Training duration (epochs)** — more epochs with cosine decay usually help until they plateau. If test accuracy is still climbing at the end, train longer.
4. **Augmentation strength** — add Random Erasing / RandAugment (Section 26) if overfitting persists; back off if training accuracy struggles to rise (over-augmentation can hurt optimization).
5. **Label smoothing** — a small final tweak (0.0–0.1) for calibration and a minor accuracy bump.

**Methodology note.** This notebook already uses a fixed class-balanced 45k/5k training/validation split and selects checkpoints by validation accuracy. Keep the official test evaluation as the final report only; if you conduct a new hyperparameter sweep, do not use its test result to select a configuration.

## 26. Optional Stronger Configurations

Keep the baseline above clean and interpretable first; layer these on **individually** and measure the effect. For each: whether it mainly helps **optimization** or **generalization**, plus the trade-off.

- **A. Mixup** — trains on convex combinations of image/label pairs. *Generalization*; strong regularizer, improves calibration; can slow convergence and lower raw train accuracy.
- **B. CutMix** — pastes a patch from one image onto another and mixes labels by area. *Generalization*; often beats Mixup on CIFAR; similar convergence-slowing caveat.
- **C. RandAugment** — samples a few random augmentation ops per image at a set magnitude. *Generalization*; excellent for reducing overfitting; too-high magnitude hurts optimization.
- **D. Random Erasing** — randomly masks a rectangle of the input. *Generalization*; cheap, composes well with crop+flip; over-aggressive erasing removes signal.
- **E. EMA of weights** — keep an exponential moving average of parameters for evaluation. *Generalization*; frequently +0.2–0.5pp for free; needs a small amount of extra memory and correct BN handling.
- **F. Stronger schedules** — cosine warm restarts (SGDR) or one-cycle. *Optimization*; can speed convergence or find better minima; adds schedule hyperparameters.
- **G. Repeated-seed training** — train 3+ seeds and report mean±std. *Neither* — it doesn't improve the model, it **quantifies variance** so your conclusions are trustworthy.
